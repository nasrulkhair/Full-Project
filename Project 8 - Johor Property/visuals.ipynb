```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("transform_johor_prop.csv")
df.head()
```

```python
# 1. Histogram of Property Prices

plt.Figure(figsize=(8,6))
sns.histplot(df["prices"], bins=30, kde=True)
plt.title("Histogram of Property Prices in Johor")
plt.xlabel("Prices")
plt.ylabel("Count")
plt.show()
```

```python
# 2. Countplot of Number of Bedrooms

sns.countplot(data = df, x="total_bedroom")
plt.title("Distribution of Property Listings by Number of Rooms")
plt.xlabel("Number of Rooms")
plt.ylabel("Count")
plt.show()
```
