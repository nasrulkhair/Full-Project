# Business Requirements Document (BRD)

## 1. Project Title
**Optimizing Travel Package Performance, Customer Satisfaction Analysis & improving Data Collection**

---

## 2. Stakeholder(s)
- **Mrs I**, General Manager
- **Sales Department**
- **Operations Team**

---

## 3. Background & Business Need
Our company offers a variety of travel packages across Malaysia and Southeast Asia. Management wants deeper insights into:
- Which packages perform best in terms of sales and satisfaction
- Which customer segments are the most profitable or risky
- How travel guides and service quality affect refund and feedback trends

---

## 4. Objectives
- Identify top-selling and high-satisfaction packages
- Understand behavior by customer segment (age, gender, type)
- Analyze feedback and refund patterns to improve future service
- Recommend improvements based on real patterns, not assumptions

---

## 5. Scope
- Analyze sales, customer, guide, refund, and feedback data for 2023–2024
- Output should include both aggregated KPIs and granular insights

---

## 6. Deliverables
- BigQuery SQL analysis queries
- Power BI dashboard with filters for customer/package dimensions
- KPI Summary page
- Feedback/refund insights by guide, package, customer
- Business recommendations document

---

## 7. Key KPIs
- Total Bookings and Revenue (by month, by channel)
- Average Satisfaction Score (by package, by guide)
- Refund Rate and Volume (by reason, by month)
- Top Customer Segments (age group, type)
- Net Promoter Score (via `WouldRecommend` %)

---

## 8. Tools & Tech Stack
- Google BigQuery (querying & transformation)
- Python (synthetic data generation)
- GCS (storage)
- Power BI (dashboard)
- GitHub (version control & documentation)
- Prezi (presentation)
