```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine

server = "NASRULKHAIR\SQLEXPRESS"
database = "Data Warehouse"

conn_str = f"mssql+pyodbc://@{server}/{database}?driver=ODBC+Driver+17+for+SQL+Server&Trusted_Connection=yes"
engine = create_engine(conn_str)
```

```python
budgets = pd.read_excel("Budgets.xlsx")

# Rename the columns to date and budget
budgets.rename(columns = {
    "Date": "date",
    "Budget": "budget"
},inplace=True)

budgets.head()
```

```python
# Calendar table

calendars = pd.read_sql("SELECT * FROM calendar", engine)
calendars.head()

# Rename the columns 
calendars.rename(columns = {
    "DateKey": "datekey",
    "Date": "date",
    "Day": "day",
    "Month": "month",
    "Year": "year",
    "MonthShort": "monthshort",
    "MonthNo": "monthno",
    "Quarter": "quarter",
    "WeekNr": "weeknr"
},inplace=True)


# Removing the columns that are not needed
calendars = calendars.drop(columns = ["datekey", "month"])

calendars["date"] = pd.to_datetime(calendars["date"], dayfirst=True, errors="coerce")
# print(calendars.dtypes)
print(calendars.head())
```

```python
# Customers Table
customers = pd.read_sql("SELECT * FROM customers", engine)
print(customers.head())

# Rename the columns 
#customers.columns = customers.columns.str.strip()
customers.rename(columns = {
    "CustomerKey": "customer_key",
    "First_Name": "first_name",
    "Last_Name": "last_name",
    "Full_Name": "full_name",
    "Gender": "gender",
    "DateFirstPurchase": "date_first_purchase",
    "Customer_City": "customer_city"
},inplace=True)

customers["date_first_purchase"] = pd.to_datetime(customers["date_first_purchase"], dayfirst=True, errors="coerce")
print(customers.dtypes)
```

```python
# Internet_Sales Table
internet_sales = pd.read_sql("SELECT * FROM internet_sales", engine)
print(internet_sales.head())

internet_sales.rename(columns = {
    "ProductKey": "product_key",
    "OrderDateKey": "order_date_key",
    "DueDateKey": "due_date_key",
    "ShipDateKey": "ship_date_key",
    "CustomerKey": "customer_key",
    "SalesOrderNumber": "sales_order_number",
    "SalesAmount": "sales_amount"
},inplace=True)

date_col = ["order_date_key", "due_date_key", "ship_date_key"]
for col in date_col:
    internet_sales[col] = pd.to_datetime(internet_sales[col], dayfirst=True, errors="coerce")
    
print(internet_sales.dtypes)
```

```python
# Products Table

products = pd.read_sql("SELECT * FROM products", engine)

products.rename(columns = {
    "ProductKey": "product_key",
    "ProductItemCode": "product_item_code",
    "Example": "example"
},inplace=True)

print(products.head())
```
